```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon
```

```python
mu = 50
sigma = 10

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
normal_pdf = norm.pdf(x, mu, sigma)
lam = 0.5

x_exp = np.linspace(0, 10, 1000)
exponential_pdf = expon.pdf(x_exp, scale=1/lam)
sample_size = 30
num_samples = 1000

samples = np.random.exponential(
    scale=1/lam,
    size=(num_samples, sample_size)
)

sample_means = np.mean(samples, axis=1)
```

```python
plt.figure(figsize=(12, 4))

# Normal Distribution
plt.subplot(1, 3, 1)
plt.plot(x, normal_pdf)
plt.title("Normal Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid()

# Exponential Distribution
plt.subplot(1, 3, 2)
plt.plot(x_exp, exponential_pdf)
plt.title("Exponential Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid()

# Central Limit Theorem
plt.subplot(1, 3, 3)
plt.hist(sample_means, bins=30, density=True)
plt.title("Central Limit Theorem")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.grid()

plt.tight_layout()
plt.savefig("normal_exponential_clt.png")
plt.show()
```
