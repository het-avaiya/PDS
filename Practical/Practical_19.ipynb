```python
import pandas as pd

data = {
    "Name": ["Dhruvi", "Hasti", "Vruti", "Khushi", "Shriya"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 72, 90, 65, 80]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
print("\nSlicing:")
print(df.iloc[0:3, 0:3])
```

```python
print("\nFiltering (Marks > 75):")
print(df[df["Marks"] > 75])
```

```python
print("\nSorting by Marks:")
print(df.sort_values("Marks", ascending=False))
```

```python
data2 = {
    "Name": ["Nidhi", "Piyu"],
    "Age": [21, 20],
    "Marks": [88, 76]
}

df2 = pd.DataFrame(data2)

combined = pd.concat([df, df2], ignore_index=True)

print("\nConcatenated Data:")
print(combined)
```

```python
print("\nAggregation:")
print("Average Marks:", combined["Marks"].mean())
print("Maximum Marks:", combined["Marks"].max())
print("Minimum Marks:", combined["Marks"].min())
```

```python
combined["Normalized Marks"] = (
    (combined["Marks"] - combined["Marks"].min()) /
    (combined["Marks"].max() - combined["Marks"].min())
)

print("\nNormalized Data:")
print(combined)
```
