```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
df = pd.read_csv("TUANDROMD.csv")
```

```python
df.head()
```

```python
df = df.dropna(how="all")
X = df.drop("Label", axis=1)
y = df["Label"]

print("\nFeatures shape:", X.shape)
print("Target shape:", y.shape)
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
model = RandomForestClassifier()

model.fit(X_train, y_train)

print("\nModel training completed.")
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy*100,"%")

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

The TUNADROMD dataset is a malware classification dataset developed to support research in malware detection.
It contains **4,465 instances and 241 features** representing permission-based and API-based characteristics.
The dataset is designed for **binary classification** of samples into **Malware** and **Goodware**.
It was presented in the **2020 IEEE 4th Conference on Information & Communication Technology (CICT)**.
The associated paper is *Malware Dataset Generation and Evaluation*, DOI: **10.1109/CICT51604.2020.9312053**.
https://archive.ics.uci.edu/dataset/813/tunadromd
