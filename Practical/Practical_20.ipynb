```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 25, 30]
y2 = [5, 15, 25, 27, 12]

subjects = ["Python", "PDS", "CN", "WAD", "SS"]
marks = [85, 75, 80, 90, 70]

students = ["A", "B", "C", "D", "E"]
scores = [70, 80, 65, 90, 75]

data = [45, 50, 55, 60, 65, 70, 70, 75, 80, 85, 90]
```

```python

# Create figure
plt.figure(figsize=(14, 10))

# 1. Line Plot
plt.subplot(2, 3, 1)
plt.plot(x, y1, marker="o", label="FirstLine")
plt.plot(x, y2, marker="o", linestyle="--", label="SecondLine")
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.legend()

# 2. Bar Chart
plt.subplot(2, 3, 2)
plt.bar(subjects, marks)
plt.title("Bar Chart")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(axis="y")

# 3. Histogram
plt.subplot(2, 3, 3)
plt.hist(data, bins=5)
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid()

# 4. Pie Chart
plt.subplot(2, 3, 4)
plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Pie Chart")

# 5. Scatter Plot
plt.subplot(2, 3, 5)
plt.scatter(students, scores)
plt.title("Scatter Plot")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.grid()

# Figure properties
plt.suptitle("Different Types of Plots Using Matplotlib", fontsize=16)
plt.tight_layout()

# Save figure
plt.savefig("all_matplotlib_plots.png")

plt.show()
```
