```python
import pandas as pd

data = {
    "Name": ["Dhruvi", "Hasti", "Vruti", "Khushi", "Shriya", "Nidhi"],
    "Age": [20, 21, None, 22, 20, 100],
    "Marks": [85, 90, 78, None, 85, 95]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
df = df.drop_duplicates()
```

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

median_age = df["Age"].median()

df["Age"] = df["Age"].apply(
    lambda x: median_age if x < lower or x > upper else x
)
```

```python
print("Cleaned Data:")
print(df)
```
