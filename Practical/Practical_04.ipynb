# Practical 4
### Aim
To write Python programs using user-defined functions with different argument and return types and to demonstrate the use of math, random and statistics modules.


### 1. Parameter with No Return

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Dhruvi", 21)
```

### 2. Parameter with Return

```python
def add(a, b):
    return a + b

x = add(10, 20)
print(x)
```

### 3. No Parameter with Return

```python
def get_number():
    a = 41
    return a

x = get_number()
print(x)
```

### 4. No Parameter with No Return

```python
def message():
    print("Python for Data Science")

message()
```

### 5. Function with Two Return Values

```python
def calculate(a, b):
    add = a + b
    multiply = a * b
    return add, multiply

x, y = calculate(10, 5)
print(x)
print(y)
```

### 6. math Module

```python
import math

print(math.sqrt(25))
print(math.pow(2, 3))
print(math.factorial(5))
print(math.ceil(4.2))
print(math.floor(4.8))
print(math.pi)
```

### 7. random Module

```python
import random

print(random.randint(1, 10))
print(random.randrange(1, 20))

a = [10, 20, 30, 40, 50]
print(random.choice(a))

random.shuffle(a)
print(a)
```

### 8. statistics Module

```python
import statistics

marks = [70, 80, 90, 75, 85, 80]

print(statistics.mean(marks))
print(statistics.median(marks))
print(statistics.mode(marks))
print(statistics.variance(marks))
print(statistics.stdev(marks))
```
