# Practical 2
### Aim
To demonstrate creation and manipulation of list, tuple, set, dictionary and string using their built-in methods.

### 1. List

```python
friends = ["Dhruvi", "Hasti", "Vruti", "Khushi"]
print(friends)
print(friends[0])
print(friends[-1])

friends.append("Jiya")
print(friends)

friends.insert(1, "Shriya")
print(friends)

friends.remove("Jiya")
print(friends)

x = friends.pop()
print(x)
print(friends)

friends.sort()
print(friends)

friends.reverse()
print(friends)

print(len(friends))
print(friends.count("Shriya"))
print(friends.index("Dhruvi"))
```

### 2. Tuple

```python
girls = ("Dhruvi", "Hasti", "Vruti", "Khushi", "Jiya")
print(girls)
print(girls[0])
print(girls[-1])
print(len(girls))
print(girls.count("Dhruvi"))
print(girls.index("Vruti"))
print(girls[1:4])
```

### 3. Set

```python
girls = {"Dhruvi", "Hasti", "Vruti", "Khushi", "Hasti"}
print(girls)

girls.add("Jiya")
print(girls)

girls.update(["Shriya", "Nidhi"])
print(girls)

girls.remove("Vruti")
print(girls)

girls.discard("Priyanshi")
print(girls)

a = {"Dhruvi", "Hasti", "Vruti", "Khushi"}
b = {"Vruti", "Khushi", "Jiya", "Shriya"}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

### 4. Dictionary

```python
student = {"name": "Dhruvi", "roll_no": 37, "age": 21, "course": "Computer Engineering"}
print(student)
print(student["name"])
print(student.get("age"))

student["city"] = "Ahmedabad"
print(student)

student["age"] = 22
print(student)

student.pop("city")
print(student)

print(student.keys())
print(student.values())
print(student.items())

student.update({"age": 21, "marks": 85})
print(student)
```

### 5. String

```python
text = "Dhruvi Hasti Vruti Khushi Jiya Shriya Nidhi Priyanshi"
print(text)
print(len(text))
print(text[0])
print(text[0:6])

print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())

print(text.replace("Dhruvi", "Dhruvi Dhameliya"))
print(text.find("Jiya"))
print(text.count("i"))

print(text.startswith("Dhruvi"))
print(text.endswith("Priyanshi"))

a = "  Dhruvi  "
print(a.strip())
print(a.lstrip())
print(a.rstrip())

words = text.split(" ")
print(words)

print("-".join(words))
```
