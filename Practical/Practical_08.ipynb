### Aim
Calculate and interpret descriptivestatistics: mean, median, mode, variance, standard deviation, range, quartiles and Iqr on a given Dataset.

### 1.Importing Pandas and load CSV

```python
import pandas as pd
df = pd.read_csv("employee.csv")
```

### 2.Mean of Age


```python
df["Age"].mean()
```

### 3. Mean Std across suitable column

```python
df["Age"].describe()
```

### 4.median of Age column

```python
df["Age"].median()
```

### 5.Mode of Age Column

```python
df["Age"].mode()
```

### 6.Range of Age

```python
max=df["Age"].max()
min=df["Age"].min()
print(f"age range of emloyee in Compny is {min} to {max}")
```

### 7.quantile of Age

```python
df["Age"].quantile(0.25)
```

```python
df["Age"].quantile(0.50)
```

```python
df["Age"].quantile(0.75)
```

### 8.IQR of Age Column

```python
IQR=df["Age"].quantile(0.75) - df["Age"].quantile(0.25)
print(f"IQR of a Age column of employee dataset is: {IQR}")
```
