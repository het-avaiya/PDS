# Practical 5
### Aim
To perform read, write and append operations on text files, CSV files and binary files using Python.

### 1. Text File - Write

```python
f = open("abc.txt", "w")
f.write("Name: Dhruvi\n")
f.write("Roll No: 37\n")
f.write("Age: 21\n")
f.close()

print("Data written")
```

### 2. Text File - Read

```python
f = open("abc.txt", "r")
data = f.read()
f.close()

print(data)
```

### 3. Text File - Append

```python
f = open("abc.txt", "a")
f.write("This is appended data.\n")
f.write("Python for Data Science\n")
f.close()

print("Data appended")
```

### 4. Read Appended Text File

```python
f = open("abc.txt", "r")
print(f.read())
f.close()
```

### 5. CSV File - Write

```python
import csv

f = open("abc.csv", "w", newline="")
writer = csv.writer(f)

writer.writerow(["Name", "Roll No", "Age"])
writer.writerow(["Ashish", 41, 21])
writer.writerow(["Rahul", 42, 21])
writer.writerow(["Amit", 43, 22])

f.close()

print("CSV file created")
```

### 6. CSV File - Read

```python
f = open("abc.csv", "r")
reader = csv.reader(f)

for row in reader:
    print(row)

f.close()
```

### 7. CSV File - Append

```python
f = open("abc.csv", "a", newline="")
writer = csv.writer(f)
writer.writerow(["Khushi", 64, 20])
f.close()

print("Data appended to CSV")
```

### 8. Read CSV After Append

```python
f = open("abc.csv", "r")
reader = csv.reader(f)

for row in reader:
    print(row)

f.close()
```

### 9. Binary File - Write

```python
f = open("abc.bin", "wb")
data =b"Hello Dhruvi"
f.write(data)
f.close()

print("Binary data written")
```

### 10. Binary File - Read

```python
f = open("abc.bin", "rb")
data = f.read()
f.close()

print(data)
```

### 11. Binary File - Append

```python
f = open("abc.bin", "ab")
f.write(b" Python")
f.close()

print("Binary data appended")
```

### 12. Read Binary File After Append

```python
f = open("abc.bin", "rb")
data = f.read()
f.close()

print(data)
```
