# Practical 7
### Aim
To connect Python with a MySQL database and perform Create, Read, Update and Delete (CRUD) operations.

### 1. Connect Python with MySQL

```python
import mysql.connector

con = mysql.connector.connect(
    host="localhost",
    user="Dhruvi",
    password="1234",
    database="studentdb"
)
cur = con.cursor()
print(con)
```

### 2. Create Table

```python
cur.execute("DROP TABLE IF EXISTS students")

cur.execute("""
CREATE TABLE students (
    id INT PRIMARY KEY,
    name VARCHAR(50),
    age INT,
    marks FLOAT
)
""")

con.commit()
```

### 3. Create - Insert Data

```python
sql = "INSERT INTO students (id, name, age, marks) VALUES (%s, %s, %s, %s)"
data = (37, "Dhruvi", 21, 85.5)

cur.execute(sql, data)
con.commit()

print("Data inserted")
```

### 4. Create - Insert More Data

```python
data = [
    (72, "Hasti", 21, 78),
    (83, "Vruti", 22, 82),
    (64, "Khushi", 20, 74)
]

cur.executemany("INSERT INTO students (id, name, age, marks) VALUES (%s, %s, %s, %s)", data)
con.commit()

print("More data inserted")
```

### 5. Read - Display All Data

```python
cur.execute("SELECT * FROM students")
rows = cur.fetchall()

for row in rows:
    print(row)
```

### 6. Read - Display Selected Data

```python
cur.execute("SELECT name, marks FROM students")
rows = cur.fetchall()

for row in rows:
    print(row)
```

### 7. Update Data

```python
sql = "UPDATE students SET marks = %s WHERE id = %s"
data = (90, 41)

cur.execute(sql, data)
con.commit()

print("Data updated")
```

### 8. Check Updated Data

```python
cur.execute("SELECT * FROM students WHERE id = 41")
print(cur.fetchone())
```

### 9. Delete Data

```python
sql = "DELETE FROM students WHERE id = %s"
data = (64,)

cur.execute(sql, data)
con.commit()

print("Data deleted")
```

### 10. Display Data After Delete

```python
cur.execute("SELECT * FROM students")
rows = cur.fetchall()

for row in rows:
    print(row)
```

### 11. Close Connection

```python
cur.close()
con.close()

print("Connection closed")
```
